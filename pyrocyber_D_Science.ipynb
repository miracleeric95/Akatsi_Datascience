# Matplotlib: Used for  Visual representation of data

```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import numpy as np
```

```python
 iris = load_iris()

x=iris.data
y=iris.target
target_names= iris.target_names

# Plot bar
plt.figure(figsize=(6,7))
unique, counts = np.unique(y,return_counts=True)
plt.bar(target_names, counts,color=plt.cm.viridis(np.linspace(0,1, len(target_names))))
plt.title("Bar plot of Iris Species")
plt.xlabel("Species")
plt.ylabel("count")
plt.show()

```

```python
# Pie chart

plt.figure(figsize=(7,5))
plt.pie(np.bincount(y),labels=target_names,autopct= '%1.1f%%', startangle=140)
plt.title("Pie chart showing iris Species")
plt.show()
```

```python
# Heatmaps (Correlation Matrix)
plt.figure(figsize=(8,6))
corr_matrix=np.corrcoef(x.T)

# plot heatmap
plt.imshow(corr_matrix,cmap="coolwarm",aspect="auto")

# add annotations
for i in range(corr_matrix.shape[0]):
    for j in range(corr_matrix.shape[1]):
        plt.text(j,i,f'{corr_matrix[i,j]:.2f}' ,ha="center", va="center", color="black")

# Add Colorbar

plt.colorbar(label="Correlation Cooeffient")

#Set labels
plt.xticks(ticks=np.arange(len(iris.feature_names)),label=iris.feature_names)
plt.yticks(ticks=np.arange(len(iris.feature_names)),label=iris.feature_names)

# Add title

plt.title("Heapmap of correlation matrix (Iris Dataset)")

plt.show()
```

```python
# Line plots
plt.figure(figsize=(8,6))
plt.plot(x[:,0],label=iris.feature_names[0])
plt.xlabel("Sample index")
plt.ylabel("Sepal Length(cm)")
plt.title("line plot of Sepal length in (cm)")
plt.legend()
plt.show()
```

```python
# Scatter plot : For visualising relationshi[ between two continues variables

plt.figure(figsize=(7,8))
for i in range(len(target_names)):
    plt.scatter(x[y==i,0],x[y==i,1],label=target_names[i])
plt.xlabel("Sepal length (cm)")
plt.ylabel("Sepal width(cm)")
plt.title("Scatter plot of Sepal length vrs sepal width")
plt.legend()
plt.show()
```

```python
# Box Plot
plt.figure(figsize=(8, 6))
colors = plt.cm.viridis(np.linspace(0, 1, len(target_names)))  # Generate colors from the 'viridis' colormap

for i, species in enumerate(target_names):
    plt.boxplot(x[y == i, 2], positions=[i], widths=0.6, patch_artist=True, boxprops=dict(facecolor=colors[i]))

plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.title('Box Plot of Petal Length (cm) by Species')
plt.xticks(ticks=np.arange(len(target_names)), labels=target_names)
plt.show()
```

```python
# Box Plot
plt.figure(figsize=(8, 6))
colors = plt.cm.viridis(np.linspace(0, 1, len(target_names)))  # Generate colors from the 'viridis' colormap

for i, species in enumerate(target_names):
    plt.boxplot(x[y == i, 2], positions=[i], widths=0.6, patch_artist=True, boxprops=dict(facecolor=colors[i]))

plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.title('Box Plot of Petal Length (cm) by Species')
plt.xticks(ticks=np.arange(len(target_names)), labels=target_names)
plt.show()
```

```python
# Histogram
plt.figure(figsize=(8, 6))
plt.hist(x[:, 0], bins=20, alpha=0.6, color='cyan', label='Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.title('Histogram of Sepal Length (cm)')
plt.legend()

plt.savefig('histogram.png')
plt.show()
```
